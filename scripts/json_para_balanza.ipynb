```python
diccionario_censo = {
    'comuna_bajo_umbral': {
        1: 'Sí',
        2: 'No'
    },
    
    'area': {
        1: 'Urbano',
        2: 'Rural'
    },
    
    'p6_fuente_agua': {
        1: 'Red pública',
        2: 'Pozo o noria',
        3: 'Camión aljibe',
        4: 'Río, vertiente, estero, canal, lago, agua lluvia, etc.',
        -99: 'No respuesta',
        'NA': 'No aplica'
    },
    
    'p7_distrib_agua': {
        1: 'Con llave dentro de la vivienda',
        2: 'Con llave dentro del sitio, pero fuera de la vivienda',
        3: 'No tiene sistema, la acarrea',
        -99: 'No respuesta',
        'NA': 'No aplica'
    }
}
```

```python
diccionario_region = {
    1: 'Tarapacá',
    2: 'Antofagasta',
    3: 'Atacama',
    4: 'Coquimbo',
    5: 'Valparaíso',
    6: "Libertador General Bernardo O'Higgins",
    7: 'Maule',
    8: 'Biobío',
    9: 'La Araucanía',
    10: 'Los Lagos',
    11: 'Aysén del General Carlos Ibáñez del Campo',
    12: 'Magallanes y de la Antártica Chilena',
    13: 'Metropolitana de Santiago',
    14: 'Los Ríos',
    15: 'Arica y Parinacota',
    16: 'Ñuble'
}
```

```python
df.replace(diccionario_censo, inplace=True)
```

```python
df.replace(diccionario_region, inplace=True)
```

```python
df
```

```python
df_conteo = df.groupby(['region', 'p6_fuente_agua']).size().reset_index(name='cantidad')
df_conteo
```

```python
df_aljibe = df_conteo[df_conteo['p6_fuente_agua'] == 'Camión aljibe']
```

```python
df_aljibe
```

```python
df_ultimate = df_aljibe.reset_index()
df_ultimate
```

```python
aaaa = df_ultimate.drop(columns = ["index", 'p6_fuente_agua'])
aaaa
```
